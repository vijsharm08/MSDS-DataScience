```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(42)
data = {
    'quiz_score': np.random.randint(10, 21, 100),
    'assignment_score': np.random.randint(30, 51, 100),
    'final_exam_score': np.random.randint(50, 101, 100),
    'study_group': np.random.choice([True, False], 100, p=[0.4, 0.6]),
    'major': np.random.choice(['Science', 'Arts', 'Engineering'], 100, p=[0.4, 0.3, 0.3])
}
df = pd.DataFrame(data)
```

```python
# 1. Histogram of Final Exam Scores
plt.figure(figsize=(8, 5))
sns.histplot(df['final_exam_score'], bins=10, kde=True)
plt.title('Distribution of Final Exam Scores')
plt.xlabel('Final Exam Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# 2. Boxplot of Final Exam Scores by Study Group
plt.figure(figsize=(8, 5))
sns.boxplot(x='study_group', y='final_exam_score', data=df)
plt.title('Final Exam Scores by Study Group Participation')
plt.xlabel('Participated in Study Group')
plt.ylabel('Final Exam Score')
plt.show()
```

```python
# 3. Bivariate Plot: Final Exam Score vs. Assignment Score (with Major as hue)
plt.figure(figsize=(10, 6))
sns.scatterplot(x='assignment_score', y='final_exam_score', hue='major', data=df)
plt.title('Final Exam Score vs. Assignment Score by Major')
plt.xlabel('Assignment Score')
plt.ylabel('Final Exam Score')
plt.show()
```

```python
# 4. Additional Visualization: Bar Graph of Average Final Exam Score by Major
plt.figure(figsize=(8, 5))
sns.barplot(x='major', y='final_exam_score', data=df, errorbar=None)
plt.title('Average Final Exam Score by Major')
plt.xlabel('Major')
plt.ylabel('Average Final Exam Score')
plt.show()
```

<b> Summary of Results and Conclusion: </b>

The visualizations suggest a potential positive correlation between study group participation and final exam performance, 
as indicated by the higher median final exam score in the study group boxplot. 
The histogram of final exam scores shows a relatively normal distribution. 
The bivariate scatter plot reveals a positive relationship between assignment scores and final exam scores, 
and also highlights that students from different majors tend to achieve varying scores, 
with <b><h>'Engineering'<b><h> students potentially performing slightly better on average in both assignments and final exams. 
The bar graph of average final exam scores by major further supports this, 
showing 'Engineering' students with the highest average score.
