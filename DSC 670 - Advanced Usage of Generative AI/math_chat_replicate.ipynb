## Define Chat Prompt

Create a prompt that includes the chat history and asks for the answer to the last question.

```python
# Define the chat history as a string
chat_history = """You are a helpful math tutor that explains answers step by step.

Q: There were nine computers in the server room. Five more computers were installed each day, from Monday to Thursday. How many computers are now in the server room?

A: There are 4 days from Monday to Thursday. 5 computers were added each day. That means in total 4 * 5 = 20 computers were added. There were 9 computers initially, so now there are 9 + 20 = 29 computers. The answer is 29.

Q: Michael had 58 golf balls. On Tuesday, he lost 23 golf balls. On Wednesday, he lost 2 more. How many golf balls did he have at the end of Wednesday?

A: Michael initially had 58 balls. He lost 23 on Tuesday, so after that he has 58 - 23 = 35 balls. On Wednesday he lost 2 more so now he has 35 - 2 = 33 balls. The answer is 33.

Q: Olivia has $23. She bought five bagels for $3 each. How much money does she have left?

A: She bought 5 bagels for $3 each. This means she spent $15. She has $8 left.

Q: When I was 6 my sister was half my age. Now I'm 70 how old is my sister?

A:"""

print("Chat prompt defined successfully!")
print(f"Prompt length: {len(chat_history)} characters")
```

## Helper Function

This function will call the Replicate API with the given prompt.

```python
def run_replicate_completion(prompt: str, model: str = "mistral") -> str:
    """
    Run Replicate model inference with the given prompt.
    
    Args:
        prompt (str): The prompt to send to the model
        model (str): The model to use ('mistral', 'llama-2-70b', 'llama-2-13b')
    
    Returns:
        str: The response content
    """
    # Available models on Replicate
    models = {
        "mistral": "mistralai/mistral-7b-instruct-v0.1",
        "llama-2-70b": "meta/llama-2-70b-chat",
        "llama-2-13b": "meta/llama-2-13b-chat",
    }
    
    model_id = models.get(model, models["mistral"])
    
    print(f"Using model: {model_id}")
    
    output = replicate.run(
        model_id,
        input={
            "prompt": prompt,
            "temperature": 0.1,
            "max_new_tokens": 200,
            "top_p": 0.9,
        }
    )
    
    # Handle different output formats from Replicate
    if isinstance(output, list):
        return "".join(output)
    return str(output)

print("Helper function defined successfully!")
```

## Run Replicate Completion

Print the chat history and generate the answer for the last question using Replicate.

```python
print("=== Recreated Chat from Listing 6.4 ===")
# Print the chat history (excluding the last incomplete A:)
lines = chat_history.split('\n')
for line in lines[:-1]:  # Exclude the last "A:" line
    print(line)

print("\n=== Generating Response for Last Question ===")
try:
    generated_answer = run_replicate_completion(chat_history, model="llama-2-13b")
    print(f"A: {generated_answer.strip()}")
except Exception as e:
    print(f"Error generating response: {e}")
    print("Please ensure your Replicate API key is set correctly.")
```
